```python
import pandas as pd 

az = pd.read_csv("../data/raw/az/A_Z_medicines_dataset_of_India.csv")
mg = pd.read_csv("../data/raw/1mg/India Medicines and Drug Info Dataset.csv")

for name, df in [("az", az), ("1mg", mg)]:
    print(name,df.shape)
    print(df.columns.tolist())
    print(df.isna().mean().round(3))
    print("---")
```

```python
comp = (az["short_composition1"].fillna("").str.strip()
        + " + "
        + az["short_composition2"].fillna("").str.strip()).str.strip(" +")
counts = comp.value_counts()
print(counts.head(20))

top500_share = counts.head(500).sum() / len(az)
print(f"top 500 type of medicines cover {top500_share: .1%} of rows")
```

```python
top = counts.index[0]                      # the aceclofenac+paracetamol string
subset = az[comp == top]
print(len(subset), "products")
print(subset["manufacturer_name"].nunique(), "manufacturers")
print(subset["price(₹)"].min(), "→", subset["price(₹)"].max())
print(subset.nsmallest(3, "price(₹)")[["name","price(₹)","pack_size_label"]])
print(subset.nlargest(3,  "price(₹)")[["name","price(₹)","pack_size_label"]])
```
